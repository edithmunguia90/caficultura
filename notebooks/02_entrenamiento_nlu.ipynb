# 02 - Entrenamiento y Evaluación del Componente NLU
**Proyecto:** Asistente Inteligente para la Caficultura en Nicaragua  
**Objetivo:** Carga del dataset de intenciones y entidades anotadas (`data/intents/intents.json`), evaluación del clasificador semántico y matriz de confusión.


```python
import sys
import os
import json
from pathlib import Path
import pandas as pd
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

sys.path.insert(0, os.path.abspath(".."))
from src.nlu.intent_classifier import IntentClassifier
from src.utils.preprocessing import normalizar_texto

```

## 1. Carga del Dataset de Intenciones Anotadas


```python
with open("../data/intents/intents.json", "r", encoding="utf-8") as f:
    intents_data = json.load(f)

print(f"Total de categorías de intención: {len(intents_data['intents'])}")
for it in intents_data["intents"]:
    print(f"- {it['name']}: {len(it['examples'])} ejemplos")

```

## 2. Instanciación y Evaluación del Clasificador NLU


```python
classifier = IntentClassifier()

y_true = []
y_pred = []

for it in intents_data["intents"]:
    intent_name = it["name"]
    for ex in it["examples"]:
        y_true.append(intent_name)
        pred, conf = classifier.detectar_intent(ex)
        y_pred.append(pred)

report = classification_report(y_true, y_pred, zero_division=0)
print(report)

```

## 3. Extracción de Entidades Agronómicas Clave
Evaluación de slots: `@sintoma_observado`, `@tipo_insumo`, `@superficie_manzanas`, `@fase_fenologica`.


```python
frases_test = [
    "Tengo manchas naranjas en hojas",
    "Necesito aplicar fertilizante urea en 3 manzanas",
    "Hay ojo de gallo en fase floracion",
    "Optimizar para 2.5 mz",
    "confirmo el plan"
]

for frase in frases_test:
    ents = classifier.extraer_entidades(frase)
    print(f"Texto: '{frase}'")
    print(f"  -> Entidades extraídas: {ents}")

```
